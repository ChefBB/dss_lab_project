```python
INPUT_FILE = f'{dataset_dir}/mb_tracks.json'
MAX_FILE_SIZE = 90 * 1024 * 1024
BLOCK_SIZE = 1000

def write_chunk(objects, index):
    filename = f'{dataset_dir}/mb_tracks_{index}.json'
    with open(filename, 'w') as f:
        json.dump(objects, f)
    print(f'Created {filename} ({os.path.getsize(filename)/1024/1024:.2f} MB)')
    return filename

def split_json_by_size():
    chunk = []
    chunk_index = 1

    # Load array in a streaming-friendly way
    with open(INPUT_FILE, 'r') as f:
        data = json.load(f)  # ok if file fits in memory; otherwise I can rewrite to stream

    # iterate through data in blocks of 1000
    for start in range(0, len(data), BLOCK_SIZE):
        block = data[start:start + BLOCK_SIZE]

        # Test what happens if we append the *whole block* to the chunk
        test_chunk = chunk + block
        estimated_bytes = len(json.dumps(test_chunk))

        # If adding the block exceeds the limit, write the current chunk
        if estimated_bytes > MAX_FILE_SIZE and chunk:
            write_chunk(chunk, chunk_index)
            chunk_index += 1
            chunk = block[:]  # new chunk begins with this block
        else:
            # Otherwise just append the block
            chunk.extend(block)

    # Write final chunk
    if chunk:
        write_chunk(chunk, chunk_index)

split_json_by_size()
```
