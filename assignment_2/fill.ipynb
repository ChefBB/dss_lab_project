```python
import json
import os

MAX_SIZE = 90 * 1024 * 1024  # 90 MB in bytes

def save_json_chunks(obj, base_path, base_name="tracks_filled"):
    """
    Splits a JSON-serializable object (usually a list) into multiple files
    under MAX_SIZE bytes by progressively adding items until full.
    """
    # If it's a dict, convert to list of key-value pairs (optional—your choice)
    if isinstance(obj, dict):
        items = list(obj.items())
    else:
        items = obj

    chunk = []
    chunk_index = 0

    for i in range(0, len(items), 100):
        # Try adding this item to the chunk
        test_chunk = chunk
        test_chunk.extend(items[i:i+100])
        encoded = json.dumps(test_chunk)

        if len(encoded) > MAX_SIZE:
            # Save previous chunk
            out_path = os.path.join(base_path, f"{base_name}_{chunk_index}.json")
            with open(out_path, "w") as f:
                json.dump(chunk, f)

            chunk_index += 1
            chunk = items[i:i+100]  # start new chunk
        else:
            chunk = test_chunk

    # Save the last chunk
    if chunk:
        out_path = os.path.join(base_path, f"{base_name}_{chunk_index}.json")
        with open(out_path, "w") as f:
            json.dump(chunk, f)

    return chunk_index + 1
```

```python
count = save_json_chunks(
    tracks_filled,
    dataset_dir,
    base_name="tracks_filled"
)

print(f"Created {count} chunk files")
```
