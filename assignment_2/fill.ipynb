```python
import json
import os

parent_dir = os.path.abspath(os.path.join(os.getcwd(), os.pardir))
dataset_dir = f'{parent_dir}/dataset'

with open(f'{dataset_dir}/mb_artists.json') as f:
    artists_retrieved = json.load(f)
    
with open(f'{dataset_dir}/artists.json') as f:
    artists_og = json.load(f)
```

```python
import integration

best_artists = integration.compute_all_matching_scores(artists_og, artists_retrieved)
```

```python
import fill

artists_filled = fill.fill_artists(artists_og=artists_og, artists_retrieved=best_artists)
```

```python
with open(f'{dataset_dir}/artists_filled.json', 'w') as f:
    json.dump(artists_filled, f)
```
