```python
import json
import os

MAX_SIZE = 90 * 1024 * 1024  # 90 MB in bytes

def save_json_chunks(obj, base_path, base_name="tracks_filled"):
    """
    Splits a JSON-serializable object (usually a list) into multiple files
    under MAX_SIZE bytes by progressively adding items until full.
    """
    # If it's a dict, convert to list of key-value pairs (optional—your choice)
    if isinstance(obj, dict):
        items = list(obj.items())
    else:
        items = obj

    chunk = []
    chunk_index = 0

    for i in range(0, len(items), 100):
        # Try adding this item to the chunk
        test_chunk = chunk
        test_chunk.extend(items[i:i+100])
        encoded = json.dumps(test_chunk)

        if len(encoded) > MAX_SIZE:
            # Save previous chunk
            out_path = os.path.join(base_path, f"{base_name}_{chunk_index}.json")
            with open(out_path, "w") as f:
                json.dump(chunk, f)

            chunk_index += 1
            chunk = items[i:i+100]  # start new chunk
        else:
            chunk = test_chunk

    # Save the last chunk
    if chunk:
        out_path = os.path.join(base_path, f"{base_name}_{chunk_index}.json")
        with open(out_path, "w") as f:
            json.dump(chunk, f)

    return chunk_index + 1
```

```python
count = save_json_chunks(
    tracks_filled,
    dataset_dir,
    base_name="tracks_filled"
)

print(f"Created {count} chunk files")
```

```python
def build_albums(track: dict) -> dict:
    """
    Starting from a track dictionary, build an album dictionary with its info.
    
    Parameters
    ----------
    track: dict
        A dictionary representing the track.
        
    Returns
    -------
    dict
        A dictionary representing the album.
    """
    album = dict()
    
    # album_name
    album['album_name'] = track['album_name'].lower() if track['album_name'] else None
    
    # album_release_date
    album['album_release_date'] = track['album_release_date']
    
    # album_type
    album['album_type'] = track['album_type']
    
    return album
```

```python
# create new ids for albums
import uuid

albums = list()
ids = list()

for track in tracks_filled:
    curr_album = build_albums(track)
    flag = False
    for album in albums:
        if (album['album_name'] == curr_album['album_name'] and
            album['album_release_date'] == curr_album['album_release_date']
        ):
            flag = True
            break
    if not flag:
        new_id = str(uuid.uuid4())
        while new_id in ids:
            new_id = str(uuid.uuid4())
        
        ids.append(new_id)
        curr_album['id_album'] = new_id
        albums.append(curr_album)
        track['new_id_album'] = new_id
        
len(albums), albums
```
