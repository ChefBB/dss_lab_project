```python
import json
import os

parent_dir = os.path.abspath(os.path.join(os.getcwd(), os.pardir))
dataset_dir = f'{parent_dir}/dataset'

with open(f'{dataset_dir}/filled/artists_filled.json') as f:
    artists = json.load(f)
```

```python
artists[0].keys()
```

```python
count = 0

for artist in artists:
    if not artist.get('latitude'):
        count += 1
        print(artist)

count
```

```python
from uuid import uuid4

new_ids = []

for artist in artists:
    new_id = str(uuid4())
    while new_id in new_ids:
        new_id = str(uuid4())
    new_ids.append(new_id)
    artist['new_id_artist'] = new_id
    
artists[0]
```

```python
with open(f'{dataset_dir}/filled/artists_filled_new_id.json', 'w') as f:
    json.dump(artists, f)
```

```python
old_ids = set()

for artist in artists:
    if artist['id_author'] not in old_ids:
        old_ids.add(artist['id_author'])
        
len(old_ids)
```

```python
with open(f'{dataset_dir}/filled/tracks_filled_new_ids.json', 'r') as f:
    tracks = json.load(f)
```

```python
tracks[0].keys()
```

```python
artist_ids_tracks = set()

for track in tracks:
    artist_ids_tracks.add(track['id_artist'])

len(artist_ids_tracks)
```

```python
for artist in artists:
    if artist['id_author'] not in artist_ids_tracks:
        print(artist)
```

```python
new_track_ids = set()

for track in tracks:
    new_id = str(uuid4())
    while new_id in new_track_ids:
        new_id = str(uuid4())
    new_track_ids.add(new_id)
    track['new_track_id'] = new_id
    for artist in artists:
        if artist['id_author'] == track['id_artist']:
            track['new_id_artist'] = artist['new_id_artist']
```

```python
with open(f'{dataset_dir}/filled/tracks_filled_new_ids.json', 'w') as f:
    json.dump(tracks, f)
```
