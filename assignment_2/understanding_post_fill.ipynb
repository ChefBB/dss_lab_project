```python
mismatches = []

for track in tracks:
    flag = False
    for participation in participations:
        if participation['new_track_id'] == track['new_track_id']:
            flag = True
    if not flag:
        mismatches.append(track)

len(mismatches)
```

```python
mismatches = []

for participation in participations:
    flag = False
    for artist in artists:
        if participation['new_id_artist'] == artist['new_id_artist']:
            flag = True
            break
    if not flag:
        mismatches.append(participation)
len(mismatches)
```

```python
participations = []

for track in tracks:
    flag = False
    for artist in artists:
        if artist['id_author'] == track['id_artist']:
            flag = True
            track['new_id_artist'] = artist['new_id_artist']
            break
    if not flag:
        print(track)
        
    
    new_participation = {
        'new_track_id': track['new_track_id'],
        'new_id_artist': track['new_id_artist'],
        'isPrimary': 1
    }
    participations.append(new_participation)

    if track.get('featured_artists'):
        featured = [clean_artist(artist) for artist in track['featured_artists'].split(',')]
        for feat in featured:
            for artist in artists:
                if artist['name'].lower() == feat:
                    participations.append({
                        'new_track_id': track['new_track_id'],
                        'new_id_artist': artist['new_id_artist'],
                        'IsPrimary': 0
                    })
```
