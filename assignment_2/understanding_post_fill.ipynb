```python
import re

def clean_artist(name):
    return re.sub(r"\s*\(.*?\)$", "", name).strip().lower()
```

```python
feat_no_match = set()
feats = []

for track in tracks:
    new_feat = []
    if track.get('featured_artists'):
        featured = [clean_artist(a) for a in track['featured_artists'].split(',')]

        for f in featured:
            f_clean = f.lower()
            flag = False

            # Check in artists
            for artist in artists:
                if SequenceMatcher(None, artist['name'].lower(), f_clean).ratio() > 0.8:
                    new_feat.append(artist['name'].lower())
                    flag = True
                    break

            # Check in artists_ret (with alias)
            if not flag:
                for i in range(len(artists_ret)):
                    # direct match
                    if SequenceMatcher(None, artists_ret[i]['name'].lower(), f_clean).ratio() > 0.8:
                        new_feat.append(artists[i]['name'].lower())
                        flag = True
                        break

                    # alias match
                    if artists_ret[i].get('alias-list'):
                        for alias in artists_ret[i]['alias-list']:
                            if SequenceMatcher(None, alias['alias'].lower(), f_clean).ratio() > 0.8:
                                new_feat.append(artists[i]['name'].lower())
                                flag = True
                                break
                            
            if not flag:
                feat_no_match.add(f_clean)
                new_feat.append(f_clean)

    # append for the whole track (moved outside loop)
    feats.append(new_feat)
```

```python
from uuid import uuid4

unique_ids = set()
for artist in artists:
    unique_ids.add(artist['new_id_artist'])
    

def build_artist_object(name: str) -> dict:
    keys = [
        'id_author', 'name', 'gender',
        'birth_date', 'birth_place', 'nationality',
        'description', 'active_start', 'active_end',
        'province', 'region', 'country', 'latitude',
        'longitude', 'type', 'active-end', 'new_id_artist'
    ]
    artist_dict = {}
    
    artist_dict['name'] = name
    new_id = str(uuid4())
    while new_id in unique_ids:
        new_id = str(uuid4())
    artist_dict['new_id_artist'] = new_id
    for key in keys:
        if not artist_dict.get(key):
            artist_dict[key] = None
            
    return artist_dict
```

```python
participations = []

for track in tracks:
    participations.append({
        'new_id_artist': track['new_id_artist'],
        'new_track_id': track['new_track_id'],
        'IsPrimary': 1
    })
    if track.get('featured_artists'):
        featured = [clean_artist(artist) for artist in track['featured_artists'].split(',')]
        for feat in featured:
            for artist in artists:
                if artist['name'].lower() == feat:
                    participations.append({
                        'new_id_artist': artist['new_id_artist'],
                        'new_track_id': track['new_track_id'],
                        'IsPrimary': 0
                    })
```
