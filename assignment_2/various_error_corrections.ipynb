```python
import os
import json

parent_dir = os.path.abspath(os.path.join(os.getcwd(), os.pardir))
dataset_dir = parent_dir + '/dataset'

with open(f'{dataset_dir}/correct_ids/tracks.json') as f:
    tracks = json.load(f)
    
with open(f'{dataset_dir}/correct_ids/dates.json') as f:
    dates = json.load(f)

len(dates)
```

```python
tracks[0].keys()
```

```python
no_match = []
match = []

for track in tracks:
    flag  = False
    for date in dates:
        if date['date_id'] == track['date_id']:
            flag = True
            match.append((track, date))
            break
    if not flag:
        no_match.append(track)
        
len(no_match), len(match)
```

```python
missing_year = []

for date in dates:
    if not date['year']:
        missing_year.append(date)

len(missing_year)
```

```python
years = []

for date in dates:
    year = date.get('year')
    if year:
        years.append(year)
        
from collections import Counter

def mode(values):
    if not values:
        raise ValueError("mode() arg is an empty list")

    counts = Counter(values)
    return counts.most_common(1)[0][0]

mode(years)
```

```python
with open(f'{dataset_dir}/tracks.json') as f:
    og_tracks = json.load(f)

len(og_tracks)
```

```python
for track, og_track in zip(tracks, og_tracks):
    track['year'] = og_track['year']
    track['month'] = og_track['month']
    track['day'] = og_track['day']
```

```python
from uuid import uuid4

dates = []
new_ids = set()

for track in tracks:
    new_date = {
        'year': None,
        'month': None,
        'day': None,
    }
    if (track.get('year') and
        track.get('month') and
        track.get('day')
    ):
        new_date = {
            'year': int(track['year']),
            'month': int(track['month']),
            'day': int(track['day']),
        }
    # only year, month
    elif (track.get('year') and
        track.get('month')
    ):
        new_date = {
            'year': int(track['year']),
            'month': int(track['month']),
            'day': None
        }
    # only year
    elif track.get('year'):
        new_date = {
            'year': int(track['year']),
            'month': None,
            'day': None
        }

    flag = False
    for date in dates:
        if (date['year'] == new_date['year'] and
            date['month'] == new_date['month'] and
            date['day'] == new_date['day']
        ):
            new_date['date_id'] = date['date_id']
            track['date_id'] = date['date_id']
            flag = True
            break
    
    if not flag:
        new_id = str(uuid4())
        while new_id in new_ids:
            new_id = str(uuid4())
        new_ids.add(new_id)
        new_date['date_id'] = new_id
        dates.append(new_date)
        track['date_id'] = new_id
```
