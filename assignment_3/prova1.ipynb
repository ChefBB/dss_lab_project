```python
import pandas as pd 
```

```python
tracks_df = pd.read_json("tracks.json")
tracks_df.head()
tracks_df.info()
```

```python
artists_df = pd.read_json("../dataset/artists.json")
artists_df.head()
artists_df.info()
```

```python
#track id duplicati 

dup_track_ids = tracks_df["id"].duplicated(keep=False).sum()
print("Numero di righe con id duplicati:", dup_track_ids)

track_id_counts = tracks_df["id"].value_counts()
multi_track_ids = track_id_counts[track_id_counts > 1]

print("\nTrack ID che compaiono più di una volta:")
print(multi_track_ids)

track_duplicates = tracks_df[tracks_df["id"].isin(multi_track_ids.index)]
track_duplicates.sort_values("id").head(20)
```

```python
#album id duplicati
dup_album_ids = tracks_df["id_album"].duplicated(keep=False).sum()
print("Numero di righe con id_album duplicati:", dup_album_ids)

album_id_counts = tracks_df["id_album"].value_counts()
multi_album_ids = album_id_counts[album_id_counts > 1]

print("\nAlbum ID che compaiono più di una volta:")
print(multi_album_ids)

album_duplicates = tracks_df[tracks_df["id_album"].isin(multi_album_ids.index)]
album_duplicates.sort_values("id_album").head(20)
```

```python
# Duplicati perfetti: righe IDENTICHE su tutte le colonne
perfect_duplicates = tracks_df[tracks_df.duplicated(keep=False)]

print("Duplicati perfetti trovati:", len(perfect_duplicates))

perfect_duplicates.head(20)
```

```python
# Prendiamo solo i track_id duplicati
track_id_counts = tracks_df["id"].value_counts()
multi_ids = track_id_counts[track_id_counts > 1].index

subset = tracks_df[tracks_df["id"].isin(multi_ids)].sort_values("id")

subset.head(20)
```

```python
#album incoerenti 
album_check = (
    tracks_df.groupby("id_album")[["album_name", "album_type", "primary_artist"]]
    .nunique()
)

problematic_albums = album_check[
    (album_check["album_name"] > 1) |
    (album_check["album_type"] > 1) |
    (album_check["primary_artist"] > 1)
]

print("\nAlbum con differenze nei campi chiave:")
problematic_albums.head(20)
```

```python
#raggruppa tutte le righe per id_album raccoglie i valori unici di album_name, primary_artist, album_type
# filtra quelli che hanno valori multipli ➜ significa “stesso album ID ma album diversi”
#stampa il risultato in modo leggibile
#Serve a dimostrare che la chiave naturale id_album è sbagliata e non può essere usata come PK.
album_check = (
    tracks_df.groupby("id_album")
    .agg({
        "album_name": lambda x: list(x.dropna().unique()),
        "primary_artist": lambda x: list(x.dropna().unique()),
        "album_type": lambda x: list(x.dropna().unique())
    })
)

# Filtra solo quelli che hanno più di un valore
problematic_albums = album_check[
    album_check.apply(lambda row: 
        len(row["album_name"]) > 1 or
        len(row["primary_artist"]) > 1 or
        len(row["album_type"]) > 1,
        axis=1
    )
]

# Stampa con il formato desiderato
for idx, row in problematic_albums.iterrows():
    print(f"id_album: {idx}")
    print(f" - album_name: {row['album_name']}")
    print(f" - primary_artist: {row['primary_artist']}")
    print(f" - album_type: {row['album_type']}")
    print()
```

```python


# Raggruppamento sui track_id problematici
track_check = (
    tracks_df.groupby("id")
    .agg({
        "title": lambda x: list(x.dropna().unique()),
        "primary_artist": lambda x: list(x.dropna().unique()),
        "album": lambda x: list(x.dropna().unique())
    })
)

# Filtra solo quelli con valori multipli (incoerenti)
problematic_tracks = track_check[
    track_check.apply(lambda row:
        len(row["title"]) > 1 or
        len(row["primary_artist"]) > 1 or
        len(row["album"]) > 1,
        axis=1
    )
]

# Stampa in formato leggibile
for idx, row in problematic_tracks.iterrows():
    print(f"track_id: {idx}")
    print(f" - title: {row['title']}")
    print(f" - primary_artist: {row['primary_artist']}")
    print(f" - album: {row['album']}")
    print()
```

```python
# Set degli artisti presenti nei brani
track_artists = set(tracks_df["id_artist"].dropna().unique())

# Set degli artisti presenti nella tabella artisti
artist_ids = set(artists_df["id_author"].dropna().unique())

# Artisti usati nelle tracce ma assenti nella tabella artist
missing_artists = track_artists - artist_ids

print("Numero di artisti usati nelle tracce MA NON presenti in artist.xml:", len(missing_artists))
list(missing_artists)[:20]
```

```python
artist_check = (
    tracks_df.groupby("id_artist")["primary_artist"]
    .nunique()
    .reset_index(name="count_primary_artist")
)

problematic_artist_ids = artist_check[artist_check["count_primary_artist"] > 1]

print("Artisti con ID incoerente:", len(problematic_artist_ids))
problematic_artist_ids.head(20)
```

```python
import pandas as pd
import re

# Estrai la colonna dei featuring
feat_series = tracks_df["featured_artists"].dropna()

# Pattern valido: solo artisti separati da virgola
valid_pattern = r'^\s*[^,]+(\s*,\s*[^,]+)*\s*$'

# Trova le righe NON valide
invalid_feat = feat_series[~feat_series.astype(str).str.match(valid_pattern)]

print("Numero di featuring NON separati correttamente da virgola:", len(invalid_feat))
invalid_feat.head(20)
```
