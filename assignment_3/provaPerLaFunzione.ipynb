```python
def normalize_country(c):
    if not c or c == "(missing)":
        return None
    
    c = c.strip().lower()

    mapping = {
        "italia": "IT",
        "it": "IT",
        "italy": "IT"
    }
    
    return mapping.get(c, c.upper())
```

```python
def country_score(artist_og, artist_mb):
    og_c = normalize_country(artist_og.get("country"))
    mb_c = normalize_country(artist_mb.get("country"))

    # CASO 1: entrambi mancanti → neutro
    if og_c is None and mb_c is None:
        return 1.0

    # CASO 2: originale mancante → neutro, non influenziamo score
    if og_c is None:
        return 1.0

    # CASO 3: MB mancante → piccola penalità
    if mb_c is None:
        return 0.7

    # CASO 4: match perfetto
    if og_c == mb_c:
        return 1.0

    # CASO 5: mismatch completo
    return 0.0
```

```python
print("=== TEST COUNTRY ===")
tests = [
    ("Italia", "IT"),
    ("Italia", "US"),
    ("Italia", None),
    (None, "IT"),
    (None, None),
]

for a,b in tests:
    og = {"country": a}
    mb = {"country": b}
    print(f"{a} vs {b} =>", country_score(og, mb))
```
