```python
import json
import os
import csv

parent_dir = os.path.abspath(os.path.join(os.getcwd(), os.pardir))
dataset_dir = f'{parent_dir}/dataset'

participations = []

with open(f'{dataset_dir}/FactParticipation.csv', 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        participations.append(row)

songs = []

with open(f'{dataset_dir}/DimSong.csv', 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        songs.append(row)

artists = []

with open(f'{dataset_dir}/DimArtist.csv', 'r') as f:
    reader = csv.DictReader(f)
    for row in reader:
        artists.append(row)
```

```python
for part in participations:
    artist_id = part['ArtistID_Original']
    song_id = part['SongID_Original']
    
    for artist in artists:
        if artist['ArtistID_Original'] == artist_id:
            for song in songs:
                if song['SongID_Original'] == song_id:
                    print(f'{song['SongTitle']}:\t\t{artist['ArtistName']}, {part['IsPrimary']}')
                    break
            break
```
