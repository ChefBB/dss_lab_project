```python
import csv
import os

CSV_DIR = r"C:\Users\Win10\OneDrive - Università degli Studi di Torino\Desktop\repo_dss\dss_lab_project\dataset\Assignment5_CSV"

dim_date = set()
fact_dates = set()

# -------- read DimDate --------
with open(os.path.join(CSV_DIR, "DimDate.csv"), "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    next(reader)
    for row in reader:
        dim_date.add(row[0])   # DateKey

# -------- read FactParticipation --------
with open(os.path.join(CSV_DIR, "FactParticipation.csv"), "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    next(reader)
    for row in reader:
        fact_dates.add(row[2])   # DateKey

# -------- check difference --------
missing = fact_dates - dim_date

print("\n=== DATEKEY MANCANTI NELLA DIMDATE ===")
print(f"Totale mancanti: {len(missing)}\n")
for d in list(missing)[:50]:   # stampiamo solo i primi 50
    print(d)

if len(missing) > 50:
    print("\n(... altri non mostrati ...)")
```

```python
import json

TRACKS = r"C:\Users\Win10\OneDrive - Università degli Studi di Torino\Desktop\repo_dss\dss_lab_project\dataset\cleaned_json\tracks_clean.json"
DATES  = r"C:\Users\Win10\OneDrive - Università degli Studi di Torino\Desktop\repo_dss\dss_lab_project\dataset\cleaned_json\dates_clean.json"

tracks = json.load(open(TRACKS, "r", encoding="utf-8"))
dates  = json.load(open(DATES,  "r", encoding="utf-8"))

date_ids_tracks = {t["date_id"] for t in tracks}
date_ids_dates  = {d["date_id"] for d in dates}

missing = date_ids_tracks - date_ids_dates

print("\n=== DATE_ID USATI NELLE TRACKS MA NON PRESENTI IN DIMDATE ===")
print("Totale mancanti:", len(missing))
print()
for m in missing:
    print(m)
```
